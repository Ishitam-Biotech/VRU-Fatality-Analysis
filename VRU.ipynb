```python
import zipfile
import pandas as pd
import os

zip_path = os.path.expanduser(
    "~/JupyterProjects/US_Accidents_March23.csv.zip"
)

# Columns actually present in your dataset
required_columns = [
    "ID",
    "Source",
    "Severity",
    "Start_Time",
    "End_Time",
    "Start_Lat",
    "Start_Lng",
    "End_Lat",
    "End_Lng",
    "Distance(mi)",
    "City",
    "County",
    "State",
    "Zipcode",
    "Weather_Condition",
    "Temperature(F)",
    "Visibility(mi)",
    "Precipitation(in)",
    "Junction",
    "Crossing",
    "Traffic_Signal",
    "Traffic_Calming",
    "Sunrise_Sunset",
    "Civil_Twilight",
    "Nautical_Twilight",
    "Astronomical_Twilight"
]

output_file = os.path.expanduser(
    "~/JupyterProjects/Project1_US_Accidents.csv"
)

chunk_size = 100000
first_chunk = True
total_rows = 0

with zipfile.ZipFile(zip_path, "r") as z:
    
    csv_name = z.namelist()[0]
    
    with z.open(csv_name) as f:
        
        for chunk in pd.read_csv(
            f,
            usecols=required_columns,
            chunksize=chunk_size
        ):
            
            chunk.to_csv(
                output_file,
                mode="w" if first_chunk else "a",
                header=first_chunk,
                index=False
            )
            
            total_rows += len(chunk)
            first_chunk = False
            
            print(f"Processed {total_rows:,} rows")

print("\nFinished!")
print(f"Output file: {output_file}")
print(f"Total rows: {total_rows:,}")
```

```python
def classify_weather(weather):
    if pd.isna(weather):
        return "Unknown"
    
    w = weather.lower()
    
    # Thunderstorm / severe weather first
    if any(x in w for x in [
        "thunder", "t-storm", "tstorm", "hail", 
        "tornado", "funnel cloud", "squall"
    ]):
        return "Thunderstorm/Severe"
    
    # Snow / ice
    elif any(x in w for x in [
        "snow", "sleet", "wintry", "ice pellet", 
        "freezing rain", "freezing drizzle"
    ]):
        return "Snow/Ice"
    
    # Rain / drizzle
    elif any(x in w for x in [
        "rain", "drizzle", "shower"
    ]):
        return "Rain/Drizzle"
    
    # Fog / haze / smoke / dust
    elif any(x in w for x in [
        "fog", "mist", "haze", "smoke", "dust", "sand"
    ]):
        return "Fog/Haze/Smoke/Dust"
    
    # Clear / fair
    elif any(x in w for x in [
        "fair", "clear"
    ]):
        return "Clear/Fair"
    
    # Cloudy conditions
    elif any(x in w for x in [
        "cloudy", "overcast", "clouds"
    ]):
        return "Cloudy"
    
    else:
        return "Unknown/Other"


df["Weather_Group"] = df["Weather_Condition"].apply(classify_weather)

print(df["Weather_Group"].value_counts())
```

```python
import numpy as np
import pandas as pd

# Keep only valid posted speed limits
valid_speed = vehicle[
    ~vehicle["VSPD_LIM"].isin([0, 98, 99])
].copy()

# Function to determine the crash-level speed limit
def get_crash_speed(x):
    counts = x.value_counts()

    # If there is a unique most common speed limit
    if counts.iloc[0] > 1 or len(counts) == 1:
        return counts.index[0]

    # If multiple speed limits occur equally often, mark as ambiguous
    if (counts == counts.iloc[0]).sum() > 1:
        return np.nan

    return counts.index[0]

crash_speed = (
    valid_speed
    .groupby("ST_CASE")["VSPD_LIM"]
    .apply(get_crash_speed)
    .reset_index()
    .rename(columns={"VSPD_LIM": "Crash_Speed_Limit"})
)

# Create binary high-speed variable
crash_speed["High_Speed"] = np.where(
    crash_speed["Crash_Speed_Limit"].isna(),
    np.nan,
    (crash_speed["Crash_Speed_Limit"] > 35).astype(float)
)

display(crash_speed.head(10))

print("\nCrash-level speed-limit summary:")
print(crash_speed["High_Speed"].value_counts(dropna=False))
```

```python
# Create Night_Unlit variable

vru_analysis["Night_Unlit"] = np.where(
    vru_analysis["LGT_CONDNAME"] == "Dark - Not Lighted",
    1,
    np.where(
        vru_analysis["LGT_CONDNAME"].isin([
            "Dark - Lighted",
            "Daylight",
            "Dusk",
            "Dawn"
        ]),
        0,
        np.nan
    )
)

print("Night_Unlit summary:")
print(vru_analysis["Night_Unlit"].value_counts(dropna=False))

print("\nLighting conditions:")
print(vru_analysis["LGT_CONDNAME"].value_counts(dropna=False))
```

```python
# Check availability of all variables needed for logistic regression

regression_data = vru_analysis[
    ["Fatal_Outcome", "High_Speed", "Night_Unlit", "VRU_Type"]
].dropna().copy()

print("Total VRU records:", len(vru_analysis))
print("Records available for regression:", len(regression_data))
print("Records excluded:", len(vru_analysis) - len(regression_data))

print("\nFatal outcome:")
print(regression_data["Fatal_Outcome"].value_counts())

print("\nHigh speed:")
print(regression_data["High_Speed"].value_counts())

print("\nNight unlit:")
print(regression_data["Night_Unlit"].value_counts())

print("\nVRU type:")
print(regression_data["VRU_Type"].value_counts())
```

```python
# Check whether High_Speed and Night_Unlit are actually independent variables

print("Cross-tabulation of High_Speed and Night_Unlit:\n")

cross_tab = pd.crosstab(
    regression_data["High_Speed"],
    regression_data["Night_Unlit"],
    margins=True
)

display(cross_tab)

print("\nProportion of Night_Unlit within each High_Speed group:\n")

cross_prop = pd.crosstab(
    regression_data["High_Speed"],
    regression_data["Night_Unlit"],
    normalize="index"
)

display(cross_prop)
```

```python
import statsmodels.api as sm

# Prepare variables
X = regression_data[["High_Speed", "Night_Unlit"]].copy()
y = regression_data["Fatal_Outcome"]

# Add intercept
X = sm.add_constant(X)

# Fit logistic regression
logit_model = sm.Logit(y, X)
result = logit_model.fit()

# Display results
print(result.summary())
```

```python
import numpy as np
import pandas as pd

# Extract coefficients and confidence intervals
coef = result.params
conf = result.conf_int()

# Convert log-odds to odds ratios
odds_ratios = np.exp(coef)
conf_odds = np.exp(conf)

# Create results table
regression_results = pd.DataFrame({
    "Coefficient": coef,
    "Odds Ratio": odds_ratios,
    "CI Lower": conf_odds[0],
    "CI Upper": conf_odds[1],
    "p-value": result.pvalues
})

# Add percentage change in odds
regression_results["% Change in Odds"] = (
    (regression_results["Odds Ratio"] - 1) * 100
)

display(regression_results)
```

```python
# Keep only the accident-level variables we need
accident_selected = accident[
    [
        "ST_CASE",
        "LATITUDE",
        "LONGITUD",
        "LGT_COND",
        "LGT_CONDNAME"
    ]
].copy()

# Merge accident information with VRU records
vru_accident = vru.merge(
    accident_selected,
    on="ST_CASE",
    how="left"
)

print("VRU records:", len(vru_accident))

print("\nMissing values:")
print(
    vru_accident[
        ["LATITUDE", "LONGITUD", "LGT_CONDNAME"]
    ].isna().sum()
)

print("\nLighting conditions:")
print(vru_accident["LGT_CONDNAME"].value_counts())
```

```python
import numpy as np

# Keep the vehicle speed-limit information we need
speed_data = vehicle[
    ["ST_CASE", "VEH_NO", "VSPD_LIM"]
].copy()

# Remove unknown / not-reported speed-limit codes
speed_data = speed_data[
    ~speed_data["VSPD_LIM"].isin([0, 98, 99])
].copy()

# Function to determine one crash-level speed limit
def get_crash_speed(x):
    counts = x.value_counts()
    max_count = counts.max()
    
    # All values with the highest frequency
    modes = counts[counts == max_count].index
    
    # Unique mode → use it
    if len(modes) == 1:
        return modes[0]
    
    # Tie between multiple speed limits → ambiguous
    return np.nan

# Calculate one speed limit per crash
crash_speed = (
    speed_data
    .groupby("ST_CASE")["VSPD_LIM"]
    .apply(get_crash_speed)
    .reset_index()
    .rename(columns={"VSPD_LIM": "Crash_Speed_Limit"})
)

# Create binary high-speed variable
crash_speed["High_Speed"] = np.where(
    crash_speed["Crash_Speed_Limit"].isna(),
    np.nan,
    (crash_speed["Crash_Speed_Limit"] > 35).astype(float)
)

print("Number of crashes with a speed value:",
      crash_speed["Crash_Speed_Limit"].notna().sum())

print("Number of crashes with ambiguous/missing speed:",
      crash_speed["Crash_Speed_Limit"].isna().sum())

print("\nHigh-speed classification:")
print(crash_speed["High_Speed"].value_counts(dropna=False))

print("\nCrash-level speed limits:")
print(crash_speed["Crash_Speed_Limit"].value_counts(dropna=False).sort_index())
```

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm

# 1. Extract valid speed limits
speed_data = vehicle[["ST_CASE", "VSPD_LIM"]].copy()

speed_data = speed_data[
    ~speed_data["VSPD_LIM"].isin([0, 98, 99])
]

# 2. Find the most common speed limit for each crash
def get_crash_speed(x):
    counts = x.value_counts()
    max_count = counts.max()
    modes = counts[counts == max_count].index
    
    if len(modes) == 1:
        return modes[0]
    else:
        return np.nan

crash_speed = (
    speed_data
    .groupby("ST_CASE")["VSPD_LIM"]
    .apply(get_crash_speed)
    .reset_index()
)

crash_speed = crash_speed.rename(
    columns={"VSPD_LIM": "Crash_Speed_Limit"}
)

# 3. Create high-speed indicator
crash_speed["High_Speed"] = np.where(
    crash_speed["Crash_Speed_Limit"].isna(),
    np.nan,
    (crash_speed["Crash_Speed_Limit"] > 35).astype(float)
)

# 4. Merge with VRU records
vru_analysis = vru_accident.merge(
    crash_speed[
        ["ST_CASE", "Crash_Speed_Limit", "High_Speed"]
    ],
    on="ST_CASE",
    how="left"
)

# 5. Create nighttime-unlit indicator
vru_analysis["Night_Unlit"] = np.where(
    vru_analysis["LGT_CONDNAME"] == "Dark - Not Lighted",
    1,
    np.where(
        vru_analysis["LGT_CONDNAME"].isin([
            "Dark - Lighted",
            "Daylight",
            "Dusk",
            "Dawn"
        ]),
        0,
        np.nan
    )
)

# 6. Create final regression dataset
regression_data = vru_analysis[
    [
        "Fatal_Outcome",
        "High_Speed",
        "Night_Unlit",
        "VRU_Type"
    ]
].dropna()

# 7. Display results
print("========================================")
print("RISK FACTOR DATASET")
print("========================================")

print("Total usable records:", len(regression_data))

print("\nFatal outcome:")
print(regression_data["Fatal_Outcome"].value_counts())

print("\nHigh speed (>35 mph):")
print(regression_data["High_Speed"].value_counts())

print("\nNight unlit:")
print(regression_data["Night_Unlit"].value_counts())
```

```python
# Logistic regression

X = regression_data[
    ["High_Speed", "Night_Unlit"]
].copy()

y = regression_data["Fatal_Outcome"]

# Add intercept
X = sm.add_constant(X)

# Fit model
result = sm.Logit(y, X).fit()

print(result.summary())
```

```python
# Calculate Odds Ratios and 95% Confidence Intervals

odds_ratios = np.exp(result.params)
conf_int = np.exp(result.conf_int())

results_table = pd.DataFrame({
    "Coefficient": result.params,
    "Odds Ratio": odds_ratios,
    "CI Lower": conf_int[0],
    "CI Upper": conf_int[1],
    "p-value": result.pvalues
})

print(results_table)
```

```python
# Odds ratios and 95% confidence intervals

or_values = np.exp(result.params)
or_ci = np.exp(result.conf_int())

or_table = pd.DataFrame({
    "Risk Factor": [
        "High Speed (>35 mph)",
        "Night + Unlit"
    ],
    "Odds Ratio": [
        or_values["High_Speed"],
        or_values["Night_Unlit"]
    ],
    "CI Lower": [
        or_ci.loc["High_Speed", 0],
        or_ci.loc["Night_Unlit", 0]
    ],
    "CI Upper": [
        or_ci.loc["High_Speed", 1],
        or_ci.loc["Night_Unlit", 1]
    ],
    "p-value": [
        result.pvalues["High_Speed"],
        result.pvalues["Night_Unlit"]
    ]
})

print(or_table.round(3))
```
